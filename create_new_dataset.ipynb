```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
```

## Load Original Training Data

```python
# Load the training data with dropped column
train_df = pd.read_csv('./dataset/train_dropped_column.csv')
print(f"Original training data shape: {train_df.shape}")
print(f"\nFirst few rows:")
train_df.head()
```

## Check Class Distribution

```python
# Check target distribution in original data
if 'Exited' in train_df.columns:
    print("Target distribution in original data:")
    print(train_df['Exited'].value_counts())
    print(f"\nPercentage distribution:")
    print(train_df['Exited'].value_counts(normalize=True) * 100)
```

## Create 20% Sample with Stratification

We'll use stratified sampling to maintain the class distribution.

```python
# Set random seed for reproducibility
np.random.seed(42)

# Create 20% sample with stratification if target column exists
if 'Exited' in train_df.columns:
    _, sample_df = train_test_split(
        train_df,
        test_size=0.2,
        stratify=train_df['Exited'],
        random_state=42
    )
else:
    # Random sampling if no target column
    sample_df = train_df.sample(frac=0.2, random_state=42)

print(f"Sample dataset shape: {sample_df.shape}")
print(f"Percentage of original: {len(sample_df) / len(train_df) * 100:.2f}%")
```

## Verify Sample Distribution

```python
# Verify target distribution in sample
if 'Exited' in sample_df.columns:
    print("Target distribution in sample:")
    print(sample_df['Exited'].value_counts())
    print(f"\nPercentage distribution:")
    print(sample_df['Exited'].value_counts(normalize=True) * 100)
```

## Save Sample Dataset

```python
# Save the sample dataset
output_path = './dataset/train_dropped_column_20pct_sample.csv'
sample_df.to_csv(output_path, index=False)
print(f"Sample dataset saved to: {output_path}")
```

## Summary

```python
print("=" * 50)
print("SAMPLING SUMMARY")
print("=" * 50)
print(f"Original dataset size: {len(train_df):,} rows")
print(f"Sample dataset size: {len(sample_df):,} rows")
print(f"Sampling percentage: 20%")
print(f"Number of features: {len(train_df.columns)}")
print(f"\nSample saved to: {output_path}")
```
